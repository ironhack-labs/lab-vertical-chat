# Vertical Chat
A sample how to build a chat for small business using:

* GPT 35
* Panel
* OpenAI


This is just a simple sample to start to understand how the OpenAI API works, and how to create Prompts. It Is really far from beign a complete solution.
We are going to introduce some interesting points:

* The roles in a conversation.
* How is the conversations’ memory preserved?

Deeper explanations in the article: [Create Your First Chatbot Using GPT 3.5, OpenAI, Python and Panel.](https://medium.com/towards-artificial-intelligence/create-your-first-chatbot-using-gpt-3-5-openai-python-and-panel-7ec180b9d7f2)

```python
#if you need an API Key from OpenAI
#https://platform.openai.com/account/api-keys

from openai import OpenAI
import os
from dotenv import load_dotenv, find_dotenv
_ = load_dotenv(find_dotenv())

OPENAI_API_KEY  = os.getenv('OPENAI_API_KEY')
```

```python
##2. Create the Chat Function

client = OpenAI(
    # This is the default and can be omitted
    api_key=OPENAI_API_KEY, #client = OpenAI(...): Creates a connection to OpenAI's API with your API key
)

def continue_conversation(messages, temperature=0): #continue_conversation() function: Sends messages to GPT-3.5 and gets a response
    response = client.chat.completions.create(
        model="gpt-3.5-turbo",
        messages=messages,
        temperature=temperature, #temperature=0: Controls randomness, # T=0 = Deterministic (same response every time), T=1 creative/ random response
    )
    #print(str(response.choices[0].message["content"]))
    return response.choices[0].message.content
```

```python
#3: Handle User Input (Core Logic)
def add_prompts_conversation(_):
    #Step 1: Get user input
    #Get the value introduced by the user
    prompt = client_prompt.value_input
    client_prompt.value = ''

    # Step 2: Add user message to conversation history
    #Append to the context the User prompt.
    context.append({'role':'user', 'content':f"{prompt}"})

    # Step 3: Get AI response
    #Get the response.
    response = continue_conversation(context)
    
    # Step 4: Add AI response to conversation history
    #Add the response to the context.
    context.append({'role':'assistant', 'content':f"{response}"})

    # Step 5: Display in UI
    #Update the panels to show the conversation.
    panels.append(
        pn.Row('User:', pn.pane.Markdown(prompt, width=600)))
    panels.append(
        pn.Row('Assistant:', pn.pane.Markdown(response, width=600)))

    return pn.Column(*panels)
```

```python
#4. Build the Chatbot & UI
#Creating the prompt
#read and understand it.
import panel as pn  # GUI

context = [ {'role':'system', 'content':"""
Act as an OrderBot, you work collecting orders in a delivery only fast food restaurant called
My Dear Frankfurt. \
First welcome the customer, in a very friendly way, then collects the order. \
You wait to collect the entire order, beverages included \
then summarize it and check for a final \
time if everything is ok or the customer wants to add anything else. \
Finally you collect the payment.\
Make sure to clarify all options, extras and sizes to uniquely \
identify the item from the menu.\
You respond in a short, very friendly style. \
The menu includes \
burger  12.95, 10.00, 7.00 \
frankfurt   10.95, 9.25, 6.50 \
sandwich   11.95, 9.75, 6.75 \
fries 4.50, 3.50 \
salad 7.25 \
Toppings: \
extra cheese 2.00, \
mushrooms 1.50 \
martra sausage 3.00 \
canadian bacon 3.50 \
romesco sauce 1.50 \
peppers 1.00 \
Drinks: \
coke 3.00, 2.00, 1.00 \
sprite 3.00, 2.00, 1.00 \
vichy catalan 5.00 \
"""} ]

#Creating the panel.
# Step 2: Initialize Panel GUI
pn.extension()

panels = []

# Step 3: Create UI components
client_prompt = pn.widgets.TextInput(value="Hi", placeholder='Enter text here…')
button_conversation = pn.widgets.Button(name="talk")

# Step 4: Connect button to function
interactive_conversation = pn.bind(add_prompts_conversation, button_conversation)

# Step 5: Build the layout
dashboard = pn.Column(
    client_prompt,
    pn.Row(button_conversation),
    pn.panel(interactive_conversation, loading_indicator=True),
)

dashboard
```

```python

report = """
## Key Findings

### Version 1: Fine Dining Bot
✓ Works well: Sophisticated tone, premium recommendations, wine pairings
✗ Issues: May be too formal, complex terminology
→ Best for: Upscale restaurants with premium positioning

### Version 2: Fast Casual Bot
✓ Works well: Quick responses, deals focus, friendly tone
✗ Issues: May miss nutrition questions, oversimplifies complex orders
→ Best for: Quick service where speed matters
---
## What I Learned

1. **System prompts shape everything** - Changing instructions completely changed bot personality
2. **Context matters** - Each bot version matched its business type perfectly
3. **Prompt engineering is critical** - Clear role, specific instructions, and style guidelines work best
4. **Potential issues:**
   - Hallucination (inventing menu items)
   - Inconsistency (changing personality mid-conversation)
   - Forgetting context in long conversations

## Best Practices for Chatbot Prompts

✓ Be specific about role and responsibility
✓ Include actual menu/data
✓ Define communication style clearly
✓ State constraints explicitly (what NOT to do)
✓ Keep prompts concise but complete
"""

print(report)
```
